```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
dataset = sns.load_dataset('titanic')
print(dataset.head())
print(dataset.shape)
```

```python
plt.figure()
sns.histplot(dataset['age'].dropna(), bins=10, kde=True)
plt.title("Distplot - Age Distribution")
plt.show()
```

```python
plt.figure()
sns.histplot(dataset['age'].dropna(), bins=10, kde=False)
plt.title("Histogram - Age (no KDE)")
plt.show()
```

```python
sns.jointplot(x=dataset['age'], y=dataset['fare'], kind='scatter')
plt.suptitle("Joint Plot - Age vs Fare (Scatter)", y=1.02)
plt.show()
```

```python
sns.jointplot(x=dataset['age'], y=dataset['fare'], kind='hex')
plt.suptitle("Joint Plot - Age vs Fare (Hex)", y=1.02)
plt.show()
```

```python
plt.figure()
sns.rugplot(x=dataset['fare'])
plt.title("Rug Plot - Fare")
plt.show()
```

```python
plt.figure()
sns.barplot(x='sex', y='age', data=dataset)
plt.title("Bar Plot - Average Age by Gender")
plt.show()
```

```python
plt.figure()
sns.barplot(x='sex', y='age', data=dataset, estimator=np.std)
plt.title("Bar Plot - Std Deviation of Age by Gender")
plt.show()
```

```python
plt.figure()
sns.countplot(x='sex', data=dataset)
plt.title("Count Plot - Passenger Gender Count")
plt.show()
```

```python
plt.figure()
sns.boxplot(x='sex', y='age', data=dataset)
plt.title("Box Plot - Age by Gender")
plt.show()
```

```python
plt.figure()
sns.boxplot(x='sex', y='age', data=dataset, hue='survived')
plt.title("Box Plot - Age by Gender & Survival")
plt.show()
```

```python
plt.figure()
sns.violinplot(x='sex', y='age', data=dataset)
plt.title("Violin Plot - Age by Gender")
plt.show()
```

```python
plt.figure()
sns.violinplot(x='sex', y='age', data=dataset, hue='survived')
plt.title("Violin Plot - Age by Gender & Survival")
plt.show()
```

```python
plt.figure()
sns.swarmplot(x='sex', y='age', data=dataset)
plt.title("Swarm Plot - Age by Gender")
plt.show()
```

```python
plt.figure()
sns.swarmplot(x='sex', y='age', data=dataset, hue='survived')
plt.title("Swarm Plot - Age by Gender & Survival")
plt.show()
```

```python
plt.figure()
corr = dataset.corr(numeric_only=True)
sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.title("Heat Map - Feature Correlation")
plt.show()
```
