```python
import matplotlib.pyplot as plt
```

```python
print("Enter data for Line 1 (Price)")
n1 = int(input("How many data points for Line 1? "))
years1, prices1 = [], []
for i in range(n1):
    years1.append(int(input(f"  Year {i+1}: ")))
    prices1.append(float(input(f"  Price {i+1}: ")))
```

```python
print("\nEnter data for Line 2 (Price)")
n2 = int(input("How many data points for Line 2? "))
years2, prices2 = [], []
for i in range(n2):
    years2.append(int(input(f"  Year {i+1}: ")))
    prices2.append(float(input(f"  Price {i+1}: ")))
```

```python
fig, ax = plt.subplots(figsize=(10, 6))

ax.plot(years1, prices1,
        linestyle='-', linewidth=2,
        marker='o', markersize=9,
        markerfacecolor='black', markeredgecolor='black', markeredgewidth=1.5,
        color='tomato', label='Price – Line 1')

ax.plot(years2, prices2,
        linestyle='--', linewidth=2,
        marker='s', markersize=9,
        markerfacecolor='black', markeredgecolor='black', markeredgewidth=1.5,
        color='royalblue', label='Price – Line 2')

ax.set_xlabel("Year",  fontsize=13, labelpad=8)
ax.set_ylabel("Price", fontsize=13, labelpad=8)
ax.set_title("Price vs Year", fontsize=16, fontweight='bold', pad=14)
ax.legend(fontsize=11)
ax.grid(True, linestyle='--', alpha=0.4)

plt.tight_layout()
plt.show()
```
