```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
iris = sns.load_dataset('iris')
print(iris.head())
print("\nShape:", iris.shape)
print("\nColumns:", iris.columns.tolist())
```

```python
print("Features and their types:")
print(iris.dtypes)
print("\nUnique species (nominal):", iris['species'].unique())
```

```python
features = ['sepal_length', 'sepal_width', 'petal_length', 'petal_width']

plt.figure(figsize=(12, 8))
for i, feature in enumerate(features):
    plt.subplot(2, 2, i+1)
    sns.histplot(iris[feature], kde=True, bins=15, color='steelblue')
    plt.title(f'Histogram - {feature}')
    plt.xlabel(feature)
    plt.ylabel('Frequency')

plt.suptitle('Histogram for Each Feature - Iris Dataset', fontsize=14)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 8))
for i, feature in enumerate(features):
    plt.subplot(2, 2, i+1)
    ax = sns.boxplot(x='species', y=feature, data=iris,
                palette={'setosa': 'steelblue', 
                         'versicolor': 'tomato', 
                         'virginica': 'orange'})
    plt.title(f'Box Plot - {feature}')
    plt.xlabel('Species')
    plt.ylabel(feature)

plt.suptitle('Box Plot for Each Feature - Iris Dataset', fontsize=14)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 8))
for i, feature in enumerate(features):
    plt.subplot(2, 2, i+1)
    sns.violinplot(x='species', y=feature, data=iris,
                palette={'setosa': 'steelblue',
                         'versicolor': 'tomato',
                         'virginica': 'orange'})
    plt.title(f'Distribution - {feature}')
    plt.xlabel('Species')
    plt.ylabel(feature)

plt.suptitle('Distribution Comparison - Iris Dataset', fontsize=14)
plt.tight_layout()
plt.show()
```

```python
print("Outlier Summary")
for feature in features:
    Q1 = iris[feature].quantile(0.25)
    Q3 = iris[feature].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    outliers = iris[(iris[feature] < lower) | (iris[feature] > upper)]
    print(f"\n{feature}:")
    print(f"  Lower bound: {lower:.2f}, Upper bound: {upper:.2f}")
    print(f"  Number of outliers: {len(outliers)}")
    if len(outliers) > 0:
        print(f"  Outlier values: {outliers[feature].values}")
```

**Observation:**

From the histograms and box plots of the Iris dataset, it is observed that the dataset has four numeric features namely sepal length, sepal width, petal length, and petal width, and one categorical feature species having three types Setosa, Versicolor, and Virginica. From the box plots, Setosa clearly has much smaller petal length and petal width compared to Versicolor and Virginica. Sepal width has the most number of outliers which can be seen as dots outside the whiskers of the box plot. Petal length and petal width show very less overlap between species, making them the most useful features to distinguish between the three species. Overall, petal features are more helpful in identifying the species compared to sepal features.
