```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
dataset = sns.load_dataset('titanic')
print(dataset.head())
print("\nShape:", dataset.shape)
print("\nColumns:", dataset.columns.tolist())
```

```python
plt.figure(figsize=(8, 6))
ax = sns.boxplot(x='sex', y='age', data=dataset, hue='survived', 
            palette={0: 'tomato', 1: 'steelblue'})

handles, labels = ax.get_legend_handles_labels()
ax.legend(handles=handles, labels=['No (0)', 'Yes (1)'], title='Survived')

plt.title("Box Plot - Age Distribution by Gender & Survival")
plt.xlabel("Gender")
plt.ylabel("Age")
plt.show()
```

```python
dataset['is_younger'] = dataset['age'].apply(
    lambda x: 'Younger' if x < 30 else 'Not Younger'
)
```

```python
plt.figure(figsize=(8, 6))
ax = sns.boxplot(x='is_younger', y='age', data=dataset, hue='survived',
            palette={0: 'tomato', 1: 'steelblue'})

handles, labels = ax.get_legend_handles_labels()
ax.legend(handles=handles, labels=['No (0)', 'Yes (1)'], title='Survived')

plt.title("Box Plot - Younger vs Not Younger & Survival")
plt.xlabel("Age Group")
plt.ylabel("Age")
plt.show()
```

```python
dataset['is_old'] = dataset['age'].apply(
    lambda x: 'Old' if x >= 50 else 'Not Old'
)
```

```python
plt.figure(figsize=(8, 6))
ax = sns.boxplot(x='is_old', y='age', data=dataset, hue='survived',
            palette={0: 'tomato', 1: 'steelblue'})

handles, labels = ax.get_legend_handles_labels()
ax.legend(handles=handles, labels=['No (0)', 'Yes (1)'], title='Survived')

plt.title("Box Plot - Old vs Not Old & Survival")
plt.xlabel("Age Group")
plt.ylabel("Age")
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.boxplot(x='sex', y='age', data=dataset, hue='survived')
plt.title("Gender vs Age - Survival Analysis")
plt.xlabel("Gender")
plt.ylabel("Age")
plt.legend(title='Survived (0=No, 1=Yes)')
plt.show()
```

**Observation:**

From the analysis, we can see that gender had a bigger effect on survival than age. Females were more likely to survive compared to males in almost every age group. Age did have some impact, as younger passengers had slightly better chances of survival, while older people had lower chances. However, the difference based on age was not very strong. The graphs also show that both survivors and non-survivors had similar age ranges, so age alone was not a clear deciding factor. Overall, gender played the most important role, while age had only a small influence on survival.
